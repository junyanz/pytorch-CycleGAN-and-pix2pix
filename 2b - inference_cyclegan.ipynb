```python
results_dir = "/home/user/data/phyusformer_data/style_transfer/inference_data_latest/synt2realUS_large_multiple_sources/test_latest/images"
masks_dir = "/home/user/data/phyusformer_data/all_combined_synthetic_datasets_V3_BM/labels"
style_transferred_images_path = [os.path.join(results_dir, x) for x in os.listdir(results_dir) if x.endswith("scan_fake.png")]
for idx,x in enumerate(style_transferred_images_path):
    flag = True
    # get the image id from the path
    image_id = x.split("/")[-1].split("_")[0]
    # image_id = np.random.randint(5000, 10000)
    # get the mask path
    assert os.path.exists(os.path.join(masks_dir, f"label_{image_id}.pkl")), "Mask path does not exist"
    mask_path = os.path.join(masks_dir, f"label_{image_id}.pkl")
    # read the mask
    # mask = pd.read_pickle(mask_path)['clean_phantom_binary']
    # mask = pd.read_pickle(mask_path)['clean_phantom_binary']
    mask = pd.read_pickle(mask_path)['clean_phantom_sound_speed']
    # mask2 = pd.read_pickle(mask_path)['clean_phantom_sound_speed']
    mask = mask.astype(np.uint8)
    if mask.max() > 1:
        flag = False
    # plt.figure(figsize=(10, 10))
    # plt.subplot(1, 2, 1)
    # plt.imshow(mask)
    # # Make the colorbar fit in aligned to the ikmage size
    # plt.colorbar(shrink=0.5)
    # plt.subplot(1, 2, 2)
    # plt.imshow(mask2)
    # plt.colorbar(shrink=0.5)
    # plt.tight_layout()
    # plt.show()
    mask = Image.fromarray(mask)
    # save the mask
    if flag:
        mask.save(os.path.join(results_dir, f"{image_id}_mask.png"))
        # break
    print(f"Saved {idx} images")
    
```
